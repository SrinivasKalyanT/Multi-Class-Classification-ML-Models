## Dataset conversion from .txt to .csv

```python
import pandas as pd

# ---------- LOAD TXT FILES ----------
X_train = pd.read_csv(
    "/Users/srinivaskalyan/Documents/Multi-Class-Classification-ML-Models/human+activity+recognition+using+smartphones/UCI HAR Dataset/train/X_train.txt",
    delim_whitespace=True,
    header=None
)

y_train = pd.read_csv(
    "/Users/srinivaskalyan/Documents/Multi-Class-Classification-ML-Models/human+activity+recognition+using+smartphones/UCI HAR Dataset/train/y_train.txt",
    header=None,
    names=["label"]
)

X_test = pd.read_csv(
    "/Users/srinivaskalyan/Documents/Multi-Class-Classification-ML-Models/human+activity+recognition+using+smartphones/UCI HAR Dataset/test/X_test.txt",
    delim_whitespace=True,
    header=None
)

y_test = pd.read_csv(
    "/Users/srinivaskalyan/Documents/Multi-Class-Classification-ML-Models/human+activity+recognition+using+smartphones/UCI HAR Dataset/test/y_test.txt",
    header=None,
    names=["label"]
)
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd

# Split 80% train, 20% validation
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train   # important for multi-class
)

print(X_tr.shape, X_val.shape)
```

```python
# Combine X and y
val_df = X_val.copy()
val_df["label"] = y_val.values

# Save
val_df.to_csv("X_val.csv", index=False)
```

```python
# ---------- SAVE SEPARATE CSVs ----------
X_train.to_csv("X_train.csv", index=False)
y_train.to_csv("y_train.csv", index=False)
X_test.to_csv("X_test.csv", index=False)
y_test.to_csv("y_test.csv", index=False)
```
